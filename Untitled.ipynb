```python
import numpy as np
import matplotlib.pyplot as plt
X=np.array([1,2,3,4,5])
Y=np.array([2,3,5,6,2])
x_mean=np.mean(X)
y_mean=np.mean(Y)
n=np.sum((X-x_mean)*(Y-y_mean))
d=np.sum((X-x_mean)**2)
b1=n/d;
b0=y_mean-b1*x_mean
y_predict=b0+b1*X
mse=np.mean((Y-y_predict)**2)
print(f"slope:{b1:.3f}")
print(f"intercept:{b0:.3f}")
print(f"mse:{mse:.3f}")
plt.scatter(X,Y,color="red",label="actual")
plt.plot(X,y_predict,color="yellow",label="regression line")
for i in range(len(X)):
    plt.vlines(X[i],Y[i],y_predict[i],colors="blue",linestyles="dashed")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("regression ")
plt.legend()
plt.show()
```
